```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# ==========================================
# 1. SETUP AND MODEL LOADING
# ==========================================
# In Google Colab, uploaded files go to the /content/ folder by default.
model_path = '/content/latest_manifold.pt'

print("📥 Loading manifold weights...")
# Load the model weights, mapping to CPU since we don't necessarily need a GPU for this
checkpoint = torch.load(model_path, map_location=torch.device('cpu'))

# ==========================================
# 2. HASH GENERATION ENGINE
# ==========================================
def generate_neural_hash(wallet_address, checkpoint_data):
    """
    Converts a wallet address into a neural hash using the manifold model weights.
    """
    print(f"🔗 Processing Wallet Address: {wallet_address}")

    # Convert the wallet address string into a numerical byte array
    input_bytes = np.frombuffer(wallet_address.encode('utf-8'), dtype=np.uint8)
    input_tensor = torch.tensor(input_bytes).float().unsqueeze(0)

    # Extract the perceptor weights to act as our hashing matrix
    weights = checkpoint_data['perceptor.weight']

    # Identify how many input features the weight matrix expects
    input_dim = weights.shape[1]

    # Pad or truncate the input tensor so it perfectly matches the model's dimensions
    if input_tensor.shape[1] < input_dim:
        # If the wallet is too short, pad it with zeros
        padding = torch.zeros((1, input_dim - input_tensor.shape[1]))
        input_tensor = torch.cat([input_tensor, padding], dim=1)
    else:
        # If it's too long, truncate it
        input_tensor = input_tensor[:, :input_dim]

    # Project the wallet data into the neural manifold space
    hash_manifold = torch.matmul(input_tensor, weights.T)

    # Apply a non-linear activation (tanh) to "squash" values between -1 and 1
    neural_hash = torch.tanh(hash_manifold)

    return neural_hash

# ==========================================
# 3. GRAPHIC TRANSLATION
# ==========================================
def hash_to_graphic(hash_tensor, save_path="/content/wallet_graphic.png"):
    """
    Translates the mathematical hash tensor into a visual graphic and displays it.
    """
    print("🎨 Translating hash to graphic...")

    # Normalize the tensor values to a 0-255 scale to represent pixel intensities
    normalized_data = ((hash_tensor - hash_tensor.min()) / (hash_tensor.max() - hash_tensor.min()) * 255)

    # Detach from PyTorch and convert to a standard 1D numpy array
    pixel_array = normalized_data.detach().numpy().astype(np.uint8)[0]

    # Determine the dimensions for a perfectly square graphic
    # We take the square root of the total features
    size = int(len(pixel_array)**0.5)

    # Crop the flat array and reshape it into a 2D grid (size x size)
    square_array = pixel_array[:size*size].reshape((size, size))

    # Create an image plot using Matplotlib
    plt.figure(figsize=(6, 6))

    # We use the 'magma' colormap for a stylized, cryptographic look.
    # You can change this to 'viridis', 'plasma', or 'gray' if you prefer!
    plt.imshow(square_array, cmap='magma')
    plt.axis('off') # Hide the axis numbers
    plt.title("Neural Wallet Graphic", fontsize=16)

    # Save the file to the Colab content folder and display it inline
    plt.savefig(save_path, bbox_inches='tight', pad_inches=0)
    plt.show()
    print(f"✅ Graphic generated and saved to {save_path}")

# ==========================================
# 4. EXECUTION
# ==========================================
# Replace this string with any wallet address you want to test
target_wallet = "0xfc1edb94e216f422224a9d903e56c72ed14dc287"

# Run the functions
resulting_hash = generate_neural_hash(target_wallet, checkpoint)
hash_to_graphic(resulting_hash)
```
